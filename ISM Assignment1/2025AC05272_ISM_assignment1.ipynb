## Question 1 — Inclusion-Exclusion Principle  [2.5 M]

**Problem:** In a survey it is found that 50% of the men like Tea, 40% like Coffee, 30% like Juice,  
20% like Tea & Coffee, 15% like Coffee & Juice, 10% like Tea & Juice, 5% like all three.  
Find the probability that a person likes **at least one** drink.

---

### Given

| Event | Probability |
|---|---|
| P(T) — Tea | 0.50 |
| P(C) — Coffee | 0.40 |
| P(J) — Juice | 0.30 |
| P(T ∩ C) | 0.20 |
| P(C ∩ J) | 0.15 |
| P(T ∩ J) | 0.10 |
| P(T ∩ C ∩ J) | 0.05 |

### Formula — Inclusion-Exclusion Principle

$$P(T \cup C \cup J) = P(T) + P(C) + P(J) - P(T\cap C) - P(C\cap J) - P(T\cap J) + P(T\cap C\cap J)$$

### Solution

$$P(T \cup C \cup J) = 0.50 + 0.40 + 0.30 - 0.20 - 0.15 - 0.10 + 0.05$$

$$= 1.20 - 0.45 + 0.05 = \boxed{0.80}$$

```python
# Question 1 — Inclusion-Exclusion Principle
P_T  = 0.50   # P(Tea)
P_C  = 0.40   # P(Coffee)
P_J  = 0.30   # P(Juice)
P_TC = 0.20   # P(Tea AND Coffee)
P_CJ = 0.15   # P(Coffee AND Juice)
P_TJ = 0.10   # P(Tea AND Juice)
P_TCJ = 0.05  # P(Tea AND Coffee AND Juice)

# Inclusion-Exclusion
P_atleast_one = P_T + P_C + P_J - P_TC - P_CJ - P_TJ + P_TCJ

print("P(T ∪ C ∪ J) = P(T) + P(C) + P(J) - P(T∩C) - P(C∩J) - P(T∩J) + P(T∩C∩J)")
print(f"             = {P_T} + {P_C} + {P_J} - {P_TC} - {P_CJ} - {P_TJ} + {P_TCJ}")
print(f"             = {P_T + P_C + P_J} - {P_TC + P_CJ + P_TJ} + {P_TCJ}")
print(f"             = {P_atleast_one}")
print(f"\nProbability that a person likes at least one drink = {P_atleast_one}")
```

## Question 2 — Binomial Distribution  [2.5 M]

**Problem:** The probability that a man hits the target is 1/3. If he fires 6 times, find the chance of hitting:  
**(i)** Exactly once  **(ii)** At least once  **(iii)** At most 5 times

---

### Given

$$p = \frac{1}{3}, \quad q = 1 - p = \frac{2}{3}, \quad n = 6$$

### Formula — Binomial Probability

$$P(X = k) = \binom{n}{k}\, p^k\, q^{n-k}$$

---

### (i) Exactly Once — P(X = 1)

$$P(X=1) = \binom{6}{1}\left(\frac{1}{3}\right)^1\left(\frac{2}{3}\right)^5
= 6 \times \frac{1}{3} \times \frac{32}{243}
= \frac{192}{729} = \frac{64}{243} \approx 0.2634$$

---

### (ii) At Least Once — P(X ≥ 1)

Using complement rule:

$$P(X \geq 1) = 1 - P(X = 0)$$

$$P(X=0) = \binom{6}{0}\left(\frac{1}{3}\right)^0\left(\frac{2}{3}\right)^6 = \left(\frac{2}{3}\right)^6 = \frac{64}{729}$$

$$P(X \geq 1) = 1 - \frac{64}{729} = \frac{665}{729} \approx 0.9122$$

---

### (iii) At Most 5 Times — P(X ≤ 5)

Using complement rule:

$$P(X \leq 5) = 1 - P(X = 6)$$

$$P(X=6) = \binom{6}{6}\left(\frac{1}{3}\right)^6\left(\frac{2}{3}\right)^0 = \left(\frac{1}{3}\right)^6 = \frac{1}{729}$$

$$P(X \leq 5) = 1 - \frac{1}{729} = \frac{728}{729} \approx 0.9986$$

```python
# Question 2 — Binomial Distribution
p = Fraction(1, 3)
q = Fraction(2, 3)
n = 6

def binomial_prob(n, k, p):
    q = 1 - p
    return Fraction(comb(n, k)) * p**k * q**(n - k)

# (i) Exactly once
P_X1 = binomial_prob(n, 1, p)
print("(i) P(X = 1) — Exactly once")
print(f"   = C(6,1) × (1/3)^1 × (2/3)^5")
print(f"   = 6 × (1/3) × (32/243)")
print(f"   = {P_X1} ≈ {float(P_X1):.4f}")

print()

# (ii) At least once
P_X0 = binomial_prob(n, 0, p)
P_atleast1 = 1 - P_X0
print("(ii) P(X ≥ 1) — At least once")
print(f"   = 1 - P(X = 0)")
print(f"   = 1 - (2/3)^6")
print(f"   = 1 - {P_X0}")
print(f"   = {P_atleast1} ≈ {float(P_atleast1):.4f}")

print()

# (iii) At most 5 times
P_X6 = binomial_prob(n, 6, p)
P_atmost5 = 1 - P_X6
print("(iii) P(X ≤ 5) — At most 5 times")
print(f"   = 1 - P(X = 6)")
print(f"   = 1 - (1/3)^6")
print(f"   = 1 - {P_X6}")
print(f"   = {P_atmost5} ≈ {float(P_atmost5):.4f}")
```

## Question 3 — Scholarship Probability (Bayes' Theorem)  [5 M]

**Problem:** A university offers scholarships.  
50% of applicants are from **urban** areas, 30% from **rural** areas, 20% are **international** students.  
A randomly selected applicant received a scholarship.

---

### Given — Prior Probabilities

$$P(\text{Urban}) = 0.50, \quad P(\text{Rural}) = 0.30, \quad P(\text{International}) = 0.20$$

$$P(\text{Urban}) + P(\text{Rural}) + P(\text{International}) = 0.50 + 0.30 + 0.20 = 1.00 \checkmark$$

### Key Assumption

The scholarship is awarded **uniformly at random** — each applicant has an equal probability of selection regardless of category:

$$P(S \mid \text{Urban}) = P(S \mid \text{Rural}) = P(S \mid \text{International}) = k$$

### Total Probability of Scholarship

$$P(S) = k \cdot P(\text{Urban}) + k \cdot P(\text{Rural}) + k \cdot P(\text{International}) = k(0.50 + 0.30 + 0.20) = k$$

---

### Formula — Bayes' Theorem

$$P(\text{Category} \mid S) = \frac{P(S \mid \text{Category}) \times P(\text{Category})}{P(S)}$$

---

### (a) P(Rural | Scholarship)

$$P(\text{Rural} \mid S) = \frac{k \times 0.30}{k} = \boxed{0.30}$$

### (b) P(Urban | Scholarship)

$$P(\text{Urban} \mid S) = \frac{k \times 0.50}{k} = \boxed{0.50}$$

### (c) Which Category Benefits Most?

| Category | P(Category) | P(Category \| Scholarship) |
|---|---|---|
| Urban | 0.50 | 0.50 |
| Rural | 0.30 | 0.30 |
| International | 0.20 | 0.20 |

**Urban** applicants benefit most in absolute terms — they constitute **50%** of scholarship recipients.

### (d) Fairness Interpretation

Since the selection is random and the posterior probabilities equal the prior probabilities, the scholarship process is **proportionally fair** — each category receives scholarships in exact proportion to its share of the applicant pool. No category is advantaged or disadvantaged relative to its representation.

However, from an **equity perspective**, rural (30%) and international (20%) students may have fewer resources than urban students, so a truly equitable process might allocate a higher proportion to these underrepresented or disadvantaged groups.

### (e) P(Rural or International | Scholarship)

Since Rural and International are **mutually exclusive** categories:

$$P(\text{Rural} \cup \text{International} \mid S) = P(\text{Rural} \mid S) + P(\text{International} \mid S)$$

$$= 0.30 + 0.20 = \boxed{0.50}$$

```python
# Question 3 — Scholarship Probability via Bayes' Theorem

# Prior probabilities
P_urban   = 0.50
P_rural   = 0.30
P_intl    = 0.20

# Equal likelihood of scholarship (random selection)
# P(S | category) = k  for all categories => P(S) = k
# By Bayes: P(category | S) = P(category)

P_rural_given_S = P_rural   # a
P_urban_given_S = P_urban   # b
P_intl_given_S  = P_intl    # derived

print("Bayes' Theorem: P(Category|S) = [P(S|Category) × P(Category)] / P(S)")
print("With uniform selection: P(S|Category) = k for all => P(Category|S) = P(Category)\n")

print(f"(a) P(Rural | Scholarship)         = {P_rural_given_S}")
print(f"(b) P(Urban | Scholarship)         = {P_urban_given_S}")
print()
print("(c) Category that benefits most:")
for cat, prob in [("Urban", P_urban_given_S), ("Rural", P_rural_given_S), ("International", P_intl_given_S)]:
    print(f"    {cat:15s}: {prob:.0%} of scholarship recipients")
print(f"    => Urban applicants benefit most ({P_urban_given_S:.0%})")
print()
print("(d) Fairness: The posterior = prior for all groups.")
print("    Each category receives scholarships proportional to its pool share.")
print("    The process is PROPORTIONALLY FAIR but may not be equitable.")
print()
P_rural_or_intl = P_rural_given_S + P_intl_given_S
print(f"(e) P(Rural ∪ International | Scholarship) = P(Rural|S) + P(International|S)")
print(f"    = {P_rural_given_S} + {P_intl_given_S} = {P_rural_or_intl}")
```
